## 2 Strings

A string is a sequence of characters inside quotes.  For example:


````python
name1 = "Urbana"
name2 = "Illinois"
````

You can print the content of a string:

````python
print(name1)
````

Or you can print its type:

````python
print(type(name1))
````

The `+` sign concatenates strings.  The `*` sign repeats a string:

````python
print(name1 + " is in " + name2*3 )
````

## 3. Numbers

Most numbers are called `floats`, or `floating point numbers`.  They are used to do calculations:

````python
x = -17.89

y = 5/3

print(x, " is less than ", y)
````

````python
print(type(y))
````

````python
print(round(y))
````

You can convert a float into a string, if you want to use the string versions of `+` and `*` instead of the numerical meanings:

````python
print("17.89*5 is ", 17.89*5, " but str(17.89)*5 is ", str(17.89)*5)
````


Integers are used for indexing.  They are numerical objects, but unlike floats, they don't have any fractional part.

````python
a = 7
b = -23
c = 0
print(type(b))
````

If you divide two integers, you usually get a float.

````python
print(b/a, " is of type", type(b/a))
````

If you want just the integer part of the division, use `//` instead.

````python
print(b//a, " is of type ", type(b//a))
````

If you convert a float to int, it will truncate the fractional part:

````python
x = - 17.89
y = 5/3
xx = int(x)
yy = int(y)
print(x, xx, y, yy)
````

## 4 "If" statements

In python, the "if" statement works like this:
* 'if', then the condition, then a colon
* The next line MUST START WITH AN INDENT!
* The first line with an indent is the end of the if statement

````python
x=3
if x==3:
    print('It worked')
````

You can check many different possibilities by using `if`, then `elif`, then `else`, as follows:

````python
x=3

if x>3:
    print("x is greater than 3!")
elif x==3:
    print("x is equal to 3!")
else:
    print("x is less than 3!")
````

## 5 Functions and modules

In python, a function is defined as follows:
* `def`, then the function name and arguments, then a colon
* The next line must be indented!
* The first line without an indent ends the function
* Use the `return` statement if you want the function to return something
* When you want to call the function, just type its name


````python
def doubleme(x):
    return 2*x

print('doubleme(4) is', doubleme(4))
print('doubleme(5) is', doubleme(5))
print('doubleme(6) is', doubleme(6))
````

A **module** in python is an external file that contains some code you want to run.  **Python is useful primarily because so many people have written so many amazing modules.** You can run their code by just downloading it, and importing it using the `import` command.  

* The `import` command runs the code in a file
* The file can be one in the same directory (a file you wrote) **or** it can be a module that's distributed with python. Some of the modules you can import are listed [here](https://docs.python.org/3/library/index.html).

For example, in this directory, there is a file called `a12am_hw1.py`.  We can import it using `import a12am_hw1`, then, **after we import it**, we can use the `help` command to find out what's inside it.

````python
import a12am_hw1
help(a12am_hw1)
````

**WARNING:** Suppose you try the following:

1. Import a file
2. Edit the file to change some stuff
3. Import it again

Using `import` again will not run the code again!  If you want to run the code again, you need to `reload` it:

````python
import importlib
importlib.reload(a12am_hw1)
help(a12am_hw1)
````
